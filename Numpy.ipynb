## Group by department ##

```python
import pandas as pd

df = pd.DataFrame({
    'EmpID': [101, 102, 103, 104],
    'Name': ['Arun', 'Priya', 'Kumar', 'Divya'],
    'Department': ['IT', 'HR', 'IT', 'Finance'],
    'Age': [25, 29, 32, 27],
    'Salary': [35000, 42000, 58000, 39000],
    'Experience': [2, 5, 8, 3],
    'City': ['Chennai', 'Madurai', 'Chennai', 'Chennai']
})

result = df.groupby('Department').agg(
    Number_of_Employees=('EmpID', 'count'),
    Average_Salary=('Salary', 'mean'),
    Highest_Salary=('Salary', 'max')
)

result = result.sort_values(
    by='Average_Salary',
    ascending=False
)

print(result)
```

## Find missing and duplicate values ##

```python

import pandas as pd

# Create the employee DataFrame
df = pd.DataFrame({
    'EmpID': [101, 102, 103, 104],
    'Name': ['Arun', 'Priya', 'Kumar', 'Divya'],
    'Department': ['IT', 'HR', 'IT', 'Finance'],
    'Age': [25, 29, 32, 27],
    'Salary': [35000, 42000, 58000, 39000],
    'Experience': [2, 5, 8, 3],
    'City': ['Chennai', 'Madurai', 'Chennai', 'Chennai']
})

df_copy = df.copy()

df_copy.loc[0, 'City'] = None

df_copy = pd.concat(
    [df_copy, df_copy.iloc[[1]]],
    ignore_index=True
)

print("DataFrame with missing value and duplicate row:")
print(df_copy)

print("\nMissing Values:")
print(df_copy.isnull().sum())

df_copy['City'] = df_copy['City'].fillna('Unknown')

print("\nAfter filling missing values:")
print(df_copy)

df_copy = df_copy.drop_duplicates()

print("\nAfter removing duplicate rows:")
print(df_copy)
```

## Use conditional logic ##

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'EmpID': [101, 102, 103, 104],
    'Name': ['Arun', 'Priya', 'Kumar', 'Divya'],
    'Department': ['IT', 'HR', 'IT', 'Finance'],
    'Age': [25, 29, 32, 27],
    'Salary': [35000, 42000, 58000, 39000],
    'Experience': [2, 5, 8, 3],
    'City': ['Chennai', 'Madurai', 'Chennai', 'Chennai']
})

df['SalaryLevel'] = np.where(
    df['Salary'] >= 50000, 'High',
    np.where(df['Salary'] >= 40000, 'Medium', 'Low')
)

print(df[['Name', 'Salary', 'SalaryLevel']])
```

## Work with a second DataFrame ##

```python

import pandas as pd

df = pd.DataFrame({
    'EmpID': [101, 102, 103, 104],
    'Name': ['Arun', 'Priya', 'Kumar', 'Divya'],
    'Department': ['IT', 'HR', 'IT', 'Finance'],
    'Age': [25, 29, 32, 27],
    'Salary': [35000, 42000, 58000, 39000],
    'Experience': [2, 5, 8, 3],
    'City': ['Chennai', 'Madurai', 'Chennai', 'Chennai']
})

dept_avg = df.groupby('Department')['Salary'].mean()

highest_dept = dept_avg.idxmax()
highest_avg = dept_avg.max()

print("Department with highest average salary:", highest_dept)
print("Highest average salary:", highest_avg)

overall_avg = df['Salary'].mean()
print("\nOverall average salary:", overall_avg)

above_avg = df[df['Salary'] > overall_avg]

print("\nEmployees earning above overall average:")
print(above_avg[['Name', 'Department', 'Salary']])
```
